```python
import pandas as pd

df = pd.read_csv("../week01/expenses.csv")
df["date"] = pd.to_datetime(df["date"])
df["month"] = df["date"].dt.strftime("%Y-%m")
df.head()
```

```python
budgets = pd.DataFrame({
    "category": ["food", "transport", "bills", "gym", 'travel'],
    "monthly_budget": [8000,3000,12000,2500,7300]
})

for how in ["inner", "left", "right", "outer"]:
    merged = df.merge(budgets, on="category", how=how)
    print(how, len(merged))
```

```python
leftist = df.merge(budgets, on="category", how="left")
print("left", len(leftist.dropna(subset=["monthly_budget"])))
```

```python
#POINT3
point_3 = df.merge(budgets, on="category", how="outer", indicator=True)
print("outer", point_3[point_3["_merge"] == "left_only"])
```

```python
#point 4
budgets_dup = pd.DataFrame({
    "category": ["food", "transport", "bills", "gym", 'travel', 'food'],
    "monthly_budget": [8000,3000,12000,2500,7300, 5000]
})
new_merge = df.merge(budgets_dup, on="category", how="left")
print(len(new_merge))
```

```python
df_1 = df.head(10)
df_2 = df.tail(10)
new_concat = pd.concat([df_1, df_2], ignore_index=True)
print(new_concat)
```

```python
tukda_2 = (
    df.merge(budgets, on="category", how="left")
      .groupby("category")
      .agg(spent=("amount", "sum"), budget=("monthly_budget", "first"))
      .assign(over=lambda d: d["spent"] > d["budget"])
)
print(tukda_2)
```
